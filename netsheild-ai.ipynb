```python
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully!")
```

```python
print("Files available in /kaggle/input:\n")

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

```python
file_path = "/kaggle/input/datasets/chethuhn/network-intrusion-dataset/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"

print("File path selected:")
print(file_path)
```

```python
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
print(df["Label"].head())
```

```python
print(df["Label"].value_counts())
```

```python
missing_values = df.isnull().sum()

print("Columns with missing values:")
print(missing_values[missing_values > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
numeric_df = df.select_dtypes(include=np.number)

infinity_count = np.isinf(numeric_df).sum().sum()

print("Total infinity values:", infinity_count)
```

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)

print("Infinity values replaced with NaN.")
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
before_dropna = df.shape

df.dropna(inplace=True)

after_dropna = df.shape

print("Before removing missing rows:", before_dropna)
print("After removing missing rows:", after_dropna)
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
before_duplicates = df.shape

df.drop_duplicates(inplace=True)

after_duplicates = df.shape

print("Before removing duplicates:", before_duplicates)
print("After removing duplicates:", after_duplicates)
```

```python
print("Labels after cleaning:")
print(df["Label"].value_counts())
```

```python
print(df.dtypes.value_counts())
```

```python
X = df.drop("Label", axis=1)
y = df["Label"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
y = y.map({
    "BENIGN": 0,
    "DDoS": 1
})

print(y.value_counts())
```

```python
non_numeric_columns = X.select_dtypes(exclude=np.number).columns

print("Non-numeric columns:")
print(non_numeric_columns.tolist())

print("\nNumber of non-numeric columns:", len(non_numeric_columns))
```

```python
print("========== DATASET SUMMARY ==========")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nFeatures:", X.shape)
print("Target:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove rows containing NaN values
df.dropna(inplace=True)

print("Dataset shape after final cleaning:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
```

```python
X = df.drop("Label", axis=1)
y = df["Label"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
y = y.map({
    "BENIGN": 0,
    "DDoS": 1
})

print("Target labels:")
print(y.value_counts())
```

```python
cleaned_file = "/kaggle/working/cicids2017_ddos_cleaned.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully!")
print(cleaned_file)
```
