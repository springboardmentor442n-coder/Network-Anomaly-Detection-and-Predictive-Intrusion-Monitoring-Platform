```python
import pandas as pd
import glob
```

```python
files = glob.glob("../data/*.csv")

print("Number of files:", len(files))

for file in files:
    print(file)
```

```python
for file in files:
    df = pd.read_csv(file, nrows=5)

    print("\nFILE:", file)
    print("Number of columns:", len(df.columns))
    print(df.columns.tolist())
```

```python
for file in files:
    df = pd.read_csv(file, nrows=1000)

    print("\nFILE:", file)
    print(df[" Label"].value_counts())
```

```python
for file in files:
    labels = pd.read_csv(file, usecols=[" Label"])

    print("\nFILE:", file)
    print(labels[" Label"].value_counts())
```

```python
for file in files:
    df = pd.read_csv(file, nrows=1)

    print(
        file,
        "→",
        len(df.columns),
        "columns"
    )
```

```python
first_columns = None

for file in files:
    df = pd.read_csv(file, nrows=1)

    if first_columns is None:
        first_columns = df.columns.tolist()

    elif df.columns.tolist() == first_columns:
        print("MATCH:", file)

    else:
        print("DIFFERENT:", file)
```

```python
for file in files:
    df = pd.read_csv(file)

    print(file)
    print("Rows:", len(df))
    print()
```

```python
import numpy as np

for file in files:

    df = pd.read_csv(file)

    print("\n" + "="*70)
    print("FILE:", file)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())

    numeric_df = df.select_dtypes(include="number")
    print("Infinite values:", np.isinf(numeric_df).sum().sum())

    del df
```
