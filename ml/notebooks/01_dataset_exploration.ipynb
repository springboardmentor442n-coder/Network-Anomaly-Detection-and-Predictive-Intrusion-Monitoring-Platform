```python
import pandas as pd
import numpy as np
import os
import glob

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Environment is ready!")
```

```python
cic_path = "../data/CICIDS2017/raw"

cic_files = glob.glob(os.path.join(cic_path, "*.csv"))

print("Number of CICIDS2017 CSV files:", len(cic_files))

for file in cic_files:
    print(os.path.basename(file))
```

```python
monday_file = os.path.join(
    cic_path,
    "Monday-WorkingHours.pcap_ISCX.csv"
)

df_monday = pd.read_csv(monday_file)

print("Dataset shape:", df_monday.shape)
print("\nColumn names:")
print(df_monday.columns.tolist())
```

```python
print("First 5 rows:")
display(df_monday.head())

print("\nData types:")
display(df_monday.dtypes)

print("\nMissing values:")
display(df_monday.isnull().sum().sort_values(ascending=False).head(15))

print("\nDuplicate rows:", df_monday.duplicated().sum())

print("\nLabel distribution:")
display(df_monday[" Label"].value_counts())
```

```python
for file in cic_files:
    print("\n" + "=" * 70)
    print(os.path.basename(file))

    labels = pd.read_csv(file, usecols=[" Label"])[" Label"]

    print("Rows:", len(labels))
    print("Labels:")
    print(labels.value_counts())
```

```python
label_summary = []

for file in cic_files:
    labels = pd.read_csv(file, usecols=[" Label"])[" Label"]
    
    counts = labels.value_counts()
    
    for label, count in counts.items():
        label_summary.append({
            "File": os.path.basename(file),
            "Label": label.strip(),
            "Count": count
        })

label_df = pd.DataFrame(label_summary)

display(label_df)
```

```python
overall_labels = label_df.groupby("Label")["Count"].sum().sort_values(ascending=False)

print("Overall CICIDS2017 Label Distribution:")
display(overall_labels.to_frame())
```
