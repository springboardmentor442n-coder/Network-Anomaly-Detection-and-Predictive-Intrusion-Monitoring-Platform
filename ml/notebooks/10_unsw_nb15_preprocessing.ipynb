```python
import pandas as pd
import numpy as np
import os

print("UNSW-NB15 preprocessing environment ready!")
```

```python
train_path = "../data/UNSW-NB15/raw/UNSW_NB15_training-set.csv"
test_path = "../data/UNSW-NB15/raw/UNSW_NB15_testing-set.csv"

print("Training file:", train_path)
print("Testing file:", test_path)
```

```python
train_df = pd.read_csv(train_path)

print("Training dataset loaded successfully!")
print("Training shape:", train_df.shape)
```

```python
test_df = pd.read_csv(test_path)

print("Testing dataset loaded successfully!")
print("Testing shape:", test_df.shape)
```

```python
print("Training columns:")
print(train_df.columns.tolist())

print("\nNumber of columns:", len(train_df.columns))
```

```python
print("Training dataset sample:")
display(train_df.head())

print("\nTesting dataset sample:")
display(test_df.head())
```

```python
print("Target-related columns:")

for column in train_df.columns:
    if column in ["label", "attack_cat"]:
        print(column)

print("\nLabel distribution:")
print(train_df["label"].value_counts())

print("\nAttack category distribution:")
print(train_df["attack_cat"].value_counts(dropna=False))
```

```python
missing_train = train_df.isnull().sum()
missing_test = test_df.isnull().sum()

print("Training missing values:")
print(missing_train[missing_train > 0])

print("\nTesting missing values:")
print(missing_test[missing_test > 0])
```

```python
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

print("Column names cleaned successfully!")
```

```python
train_df["attack_cat"] = train_df["attack_cat"].fillna("Normal")
test_df["attack_cat"] = test_df["attack_cat"].fillna("Normal")

train_df["attack_cat"] = train_df["attack_cat"].astype(str).str.strip()
test_df["attack_cat"] = test_df["attack_cat"].astype(str).str.strip()

print("Attack categories cleaned successfully!")

print("\nTraining attack categories:")
print(train_df["attack_cat"].value_counts())
```

```python
train_df = train_df.drop(columns=["id"], errors="ignore")
test_df = test_df.drop(columns=["id"], errors="ignore")

print("Identifier column removed.")
print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
print("Categorical columns:")

categorical_columns = train_df.select_dtypes(
    include=["object"]
).columns.tolist()

print(categorical_columns)

print("\nNumeric columns:")
print(len(train_df.select_dtypes(include=[np.number]).columns))
```

```python
categorical_features = [
    "proto",
    "service",
    "state"
]

existing_categorical = [
    column for column in categorical_features
    if column in train_df.columns
]

print("Categorical features to encode:")
print(existing_categorical)

train_encoded = pd.get_dummies(
    train_df,
    columns=existing_categorical,
    dtype=int
)

test_encoded = pd.get_dummies(
    test_df,
    columns=existing_categorical,
    dtype=int
)

print("\nEncoding completed!")
print("Training shape:", train_encoded.shape)
print("Testing shape:", test_encoded.shape)
```

```python
target_columns = ["label", "attack_cat"]

feature_columns = [
    column for column in train_encoded.columns
    if column not in target_columns
]

train_features = train_encoded[feature_columns].copy()
test_features = test_encoded.drop(
    columns=target_columns,
    errors="ignore"
).copy()

train_features, test_features = train_features.align(
    test_features,
    join="left",
    axis=1,
    fill_value=0
)

print("Training feature shape:", train_features.shape)
print("Testing feature shape:", test_features.shape)
```

```python
train_features = train_features.replace(
    [np.inf, -np.inf],
    np.nan
)

test_features = test_features.replace(
    [np.inf, -np.inf],
    np.nan
)

train_features = train_features.fillna(0)
test_features = test_features.fillna(0)

print("Missing and infinite values handled.")

print("Training missing values:", train_features.isnull().sum().sum())
print("Testing missing values:", test_features.isnull().sum().sum())
```

```python
y_train = train_encoded["label"].astype(int)
y_test = test_encoded["label"].astype(int)

print("Binary labels prepared!")

print("\nTraining labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())

print("\nLabel meaning:")
print("0 = NORMAL")
print("1 = ATTACK")
```

```python
train_ml_ready = train_features.copy()
train_ml_ready["Binary_Label"] = y_train.values

test_ml_ready = test_features.copy()
test_ml_ready["Binary_Label"] = y_test.values

print("ML-ready training dataset:", train_ml_ready.shape)
print("ML-ready testing dataset:", test_ml_ready.shape)
```

```python
output_dir = "../data/UNSW-NB15/ml_ready"

os.makedirs(output_dir, exist_ok=True)

train_output = os.path.join(
    output_dir,
    "UNSW_NB15_training_ml_ready.csv"
)

test_output = os.path.join(
    output_dir,
    "UNSW_NB15_testing_ml_ready.csv"
)

train_ml_ready.to_csv(
    train_output,
    index=False
)

test_ml_ready.to_csv(
    test_output,
    index=False
)

print("UNSW-NB15 preprocessing completed successfully!")

print("\nTraining saved to:")
print(train_output)

print("\nTesting saved to:")
print(test_output)
```
