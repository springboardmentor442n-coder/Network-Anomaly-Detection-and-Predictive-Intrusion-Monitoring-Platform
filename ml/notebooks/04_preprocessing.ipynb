```python
import pandas as pd
import numpy as np
import os
import glob

print("Preprocessing environment ready!")
```

```python
processed_path = "../data/CICIDS2017/processed"

processed_files = glob.glob(
    os.path.join(processed_path, "*.csv")
)

print("Number of processed files:", len(processed_files))

for file in processed_files:
    print(os.path.basename(file))
```

```python
monday_file = os.path.join(
    processed_path,
    "Monday-WorkingHours.pcap_ISCX.csv"
)

df = pd.read_csv(monday_file)

print("Dataset shape:", df.shape)
print("Number of columns:", len(df.columns))
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
X = df.drop(columns=["Label"])
y = df["Label"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
non_numeric_columns = X.select_dtypes(
    exclude=np.number
).columns

print("Non-numeric columns:")
print(non_numeric_columns.tolist())
```

```python
for column in non_numeric_columns:
    print(
        column,
        "→",
        X[column].nunique(),
        "unique values"
    )
```

```python
columns_to_drop = [
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp"
]

X = X.drop(
    columns=columns_to_drop,
    errors="ignore"
)

print("New feature shape:", X.shape)
```

```python
print(X.dtypes.value_counts())
```

```python
print(
    "Non-numeric columns remaining:",
    X.select_dtypes(exclude=np.number).columns.tolist()
)
```

```python
infinite_values = np.isinf(X).sum().sum()

print("Total infinite values:", infinite_values)
```

```python
missing_values = X.isnull().sum().sum()

print("Total missing values:", missing_values)
```

```python
print("Final feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nMissing values:", X.isnull().sum().sum())
print("Infinite values:", np.isinf(X).sum().sum())

print("\nData types:")
print(X.dtypes.value_counts())
```
